```python
"""
interest rates, fx rate with usd
interest rate differential, carry trade return
formulas:

carry trade return = interest rate differential - change in fx rate
interest rate differential = interest rate of foreign currency - interest rate of domestic currency
change in fx rate = (fx rate at time t - fx rate at time t-1) / fx rate at time t-1

"""

import json
import pathlib

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import statsmodels.api as sm
from scipy import stats
from itertools import combinations

df = pd.read_excel("/Users/partha/Documents/FX Carry/Macro Excel.xlsx")
df.head()
```

```python
usd_rate = df.loc[df["Country"] == "United States", "Interest rate (%)"].iloc[0]
df["Interest Rate Differential (%)"] = df["Interest rate (%)"] - usd_rate

usd_rate = df.loc[df["Country"] == "United States", "FX: units per 1 USD (live)"].iloc[0]
df["change in fx rate"] = (df["FX: units per 1 USD (live)"] - usd_rate)/usd_rate
```

```python
df.head(50)
```

```python
df['carry_trade_return'] = (df['Interest Rate Differential (%)'] - df['change in fx rate']) * 100
df.to_excel('/Users/partha/Documents/FX Carry/carry_trade_returns.xlsx', index=False)
    
```
